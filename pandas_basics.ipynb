# Pivot Tables

```python
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
titanic
```

```python
titanic["who"].value_counts()
```

```python
titanic.pivot_table(index= "who",columns= "class" ,values= "age" ,aggfunc= "mean")
```

```python
# the percentage of survived females and males
titanic.pivot_table(index= "sex",values= "survived" ,aggfunc= "mean")
```

```python
titanic.pivot_table(index= "who",columns= "class" ,values=[ "age", "survived" ],aggfunc= "count")
```

```python
titanic.pivot_table(index= "who",columns= "class" ,values=[ "age", "survived" ], aggfunc= {"survived": "count" , "age":"mean"})
```

```python
titanic.shape
```

```python
titanic.isnull().sum()
```

```python
titanic["age"] = titanic["age"].fillna(titanic["age"].median())
```

```python
titanic.isnull().sum()
```

```python
titanic["embarked"] = titanic["embarked"].fillna(titanic["embarked"].mode()[0])
```

```python
titanic["embark_town"] = titanic["embark_town"].fillna(titanic["embark_town"].mode()[0])
```

```python
titanic = titanic.drop(columns=["deck"])
```
